# 🐍 Python Data Types
**Категории типов:**
- Text: `str`
- Numeric: `int`, `float`, `complex`
- Sequence: `list`, `tuple`, `range`
- Mapping: `dict`
- Set: `set`, `frozenset`
- Boolean: `bool`
- Binary: `bytes`, `bytearray`, `memoryview`
- None Type: `NoneType`

**Подсказка:** используйте `type(x)` для проверки типа. Тип присваивается при создании значения, а также может задаваться конструкторами (`int()`, `list()` и т.д.).

### Задание 20. Типизация по конструктору

```python
s = str('asdasd')
i = int(123)
f = float(3.5)
c = complex(1+2J)
l = list((1, 2))
t = tuple((1, 2))
r = range(1, 4)
d = dict(one=1, two=2)
s1 = set((1, 2, 3))
f1 = frozenset((1, 2, 3))
b = bool(2)
b1 = bytes(1)
b2 = bytearray(2)
m = memoryview(bytes(2))
value = [s, i, f, c, l, t, r, d, s1, f1, b, b1, b2, m]
for i in value:
    print(i, type(i))
```
