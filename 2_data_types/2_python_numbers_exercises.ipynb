# 🐍 Python Numbers 
## Краткая теория
- Числовые типы: **int**, **float**, **complex**.
- Проверка типа: `type(x)`.
- Преобразование типов: `int()`, `float()`, `complex()` (нельзя преобразовать `complex` к `int/float`).
- Научная запись: `1e3 == 1000.0`.
- Модуль `random`: `randint`, `randrange`, `uniform`, `random`.
- Полезные функции: `round`, `abs`, `pow`, `math.sqrt`, `math.floor`, `math.ceil`.
- Комплексные числа: `a+bj`, `z.real`, `z.imag`, `abs(z)`.

### Задание 17. Infinity и NaN
Используя `float('inf')`, `float('-inf')`, `float('nan')`, покажите их поведение при сравнениях и операциях.

```python
# Ваш код здесь
import math

inf = float('inf')
neg_inf = float('-inf')
nan = float('nan')

print(f"Is inf infinite? {math.isinf(inf)}")

print(f"Is nan equal to itself? {nan == nan}")
print(f"Correct check for NaN: {math.isnan(nan)}")

print(f"inf + neg_inf = {inf + neg_inf}") 
print(f"1 / inf = {1 / inf}")
```
