## Краткая теория
- **Группы операторов:** арифметические, присваивания, сравнения, логические, идентичности, принадлежности, побитовые.
- **Приоритет:** `()` → `**` → унарные `+ - ~` → `* / // %` → `+ -` → сдвиги `<< >>` → `&` → `^` → `|` → сравнения/`is`/`in` → `not` → `and` → `or`.

- Побитовые операции работают с целыми числами и их двоичным представлением.
