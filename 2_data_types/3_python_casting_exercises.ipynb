# 🐍 Python Casting
## Краткая теория
- **int(x)** — создаёт целое число из `int`, `float` (усечение к нулю) или строки с целым.
- **float(x)** — создаёт число с плавающей точкой из `int`, `float` или строки с числом.
- **str(x)** — создаёт строку из почти любого объекта.
- Пробелы по краям строк допустимы: `int(' 42 ')`.
- Некорректные строки приводят к `ValueError`.
- Часто нужно комбинировать: `int(float('7.0'))`.

### Задание 6. Невалидные строки
Поймайте ошибку для `int('3.14')` и `float('3,14')` с помощью `try/except` и выведите тип ошибки.

```python
# Ваш код здесь#
try:
    number = int('3.14')
except ValueError as e:
    print(f"Ошибка для int('3.14'): {type(e).__name__} — {e}")

try:
    number = float('3,14')
except ValueError as e:
    print(f"Ошибка для float('3,14'): {type(e).__name__} — {e}")
```

### Задание 8. Сумма чисел в строке
