## Краткая теория
- Регистр: `upper()`, `lower()`, `casefold()`.
- Пробелы: `strip()`, `lstrip()`, `rstrip()`; `strip(chars)` удаляет указанные символы по краям.
- Замены: `replace(old, new[, count])`.
- Разбиение: `split(sep=None, maxsplit=-1)`, `splitlines()`.
- Форматирование/выравнивание: `center(width[, fillchar])`, `ljust`, `rjust`; `startswith`, `endswith`.
- Методы можно **комбинировать**: `s.strip().lower().replace(...)`.

### Задание 11. lstrip()/rstrip()

### Задание 17. center()/ljust()/rjust()

```python
# Ваш код здесь
s ="hi"
print(s.center(6,"-"))
print(s.ljust(6,"-"))
print(s.rjust(6,"-"))
```

### Задание 20. Практика: извлечение
Пусть s = 'id=12345; name=Alice; role=admin'. Получите список значений ['12345','Alice','admin'].

```python
# Ваш код здесь
s = "id=12345; name=Alice; role=admin."

lstp = []
for stp in s.split(";"):
    lstp.append(stp.split("=")[1])

print(lstp)



print([stp.split('=')[1] for stp in s.split(';')])

    
```
