## Краткая теория
- Срез: `s[start:end]` — возвращает подстроку с `start` по `end-1`.
- `s[:n]` — от начала до `n` (не включая `n`), `s[n:]` — от `n` до конца.
- Отрицательные индексы отсчитываются с конца: `-1` — последний символ.
- Шаг: `s[::k]`; реверс — `s[::-1]`.
