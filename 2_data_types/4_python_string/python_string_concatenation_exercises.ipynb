## Краткая теория
- Конкатенация строк: `a + b`.
- Добавление пробелов/символов: `a + ' ' + b`, `a + '-' + b`.
- Повтор строк: `s * n`.
- Смешивание типов требует преобразования: `str(x)` или форматирования `f'...'`, `'{}'.format(...)`.
- Для списков строк используйте `''.join(list_of_str)` — эффективно и удобно.

### Задание 7. Конкатенация в цикле
Соберите строку из символов списка ['P','y','t','h','o','n'] в одно слово с помощью конкатенации и выведите.

```python
# Ваш код здесь
sps = ['P','y','t','h','o','n']
new_spis = ""

for new in sps:
    new_spis += new

print(new_spis)
```

### Задание 8. join как альтернатива
То же самое, но используйте ''.join([...]). Объясните в комментарии, почему это предпочтительно для больших списков.

```python
# Ваш код здесь
sps = "".join(['P','y','t','h','o','n'])
print(sps)
```

### Задание 19. Форматирование: f-string
Пусть first='Ada', last='Lovelace', year=1843. Соберите: 'Ada Lovelace, 1843' с f-строкой.

```python
# Ваш код здесь
first, last, year = "Ada", "Lovelace", 1843
print(f"{first} {last}, {year}")
```

### Задание 20. Форматирование: format()
То же самое задание, но с методом .format().

```python
# Ваш код здесь
first, last, year = "Ada", "Lovelace", 1843
new_str = "{f} {l}, {y}".format(f = first, l = last,y = year)
print(new_str)
```
