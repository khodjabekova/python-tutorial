# 🐍 Python Strings

## Краткая теория
- Строки в Python заключаются в одинарные или двойные кавычки. Многострочные — в тройные кавычки.
- Внутренние кавычки разрешены, если не совпадают с ограничивающими.
- Строка — последовательность символов (индексация с 0, отрицательная — с конца).
- `for` позволяет итерироваться по символам строки.
- `len(s)` — длина строки; проверка подстроки: `'sub' in s`, `'sub' not in s`.
- Срезы: `s[start:stop:step]`, поддерживаются отрицательные индексы и шаг.
- Полезные методы: `lower`, `upper`, `strip`, `find`, `count`, и др.
