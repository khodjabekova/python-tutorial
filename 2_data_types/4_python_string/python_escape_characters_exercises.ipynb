## Краткая теория
- Escape-последовательность начинается с обратного слеша `\` и задаёт спецсимвол: `\n`, `\t`, `\r`, `\'`, `\"`, `\\` и т.д.
- Шестнадцатеричные/октальные коды: `\xhh`, `\ooo`.
- Сырые строки: `r"..."` не интерпретируют `\` как начало escape (кроме закрывающей кавычки). Удобно для путей и regex.
- Для выводимых фигурных скобок в форматировании используйте удвоение: `{{` и `}}`.
