## Краткая теория
- **Создание строк:** одинарные/двойные кавычки; многострочные — тройные кавычки.
- **Индексация/срезы:** `s[i]`, `s[a:b:c]`, отрицательные индексы, реверс `s[::-1]`.
- **Конкатенация/повтор:** `a + b`, `' '.join(list)`, `s * n`.
- **Форматирование:** f-строки (`f"{x:.2f}"`), `str.format`, ширина/выравнивание/проценты.
- **Экранирование и спецсимволы:** `\n`, `\t`, `\"`, `\\`, `\xhh`, `\uXXXX`; `r"..."` — raw-строки.
- **Методы строк:** `strip/lstrip/rstrip`, `upper/lower/title/swapcase`, `find/rfind/count`,
  `startswith/endswith`, `split/rsplit/splitlines`, `join`, `replace`, `partition/rpartition`,
  проверки `is*` (`isalpha`, `isdigit`, `isalnum`, `isspace`, `isidentifier` и т.д.).
- **Неизменяемость:** строки неизменяемы → методы возвращают **новые** строки.

### Задание 13. Выравнивание
Пусть n=42. Выведите его в поле шириной 6: справа {:>6}, слева {:<6}, по центру {:^6}.

```python
# Ваш код здесь
n = str(42)
print(n.center(6,"-"), n.ljust(6,"-"), n.rjust(6,"-"), sep = "\n")
```

### Задание 30. translate/maketrans
