## Краткая теория
- Нельзя конкатенировать строку и число без преобразования: используйте **f-строки** или `format()`.
- f-строки: `f'The price is {price}'`, модификаторы: `{price:.2f}`, ширина/выравнивание: `{:>6}`.
- `format()`: позиционные `{0}`, именованные `{name}`, методы `:,`, `:+d`, `:05d`, `:.2%`, `:x`, `:b` и др.
- Внутри placeholder можно писать выражения и вызывать функции.
