# 🐍 Python: Global Variables

## 📘 Теория: Глобальные переменные
**Глобальные переменные** — это переменные, объявленные вне функций. Они доступны в любой части программы.

```python
x = 'awesome'

def myfunc():
    print('Python is ' + x)

myfunc()
```
✅ Вы можете использовать глобальные переменные как внутри, так и вне функций.

### 🔒 Локальные переменные
Если вы создаёте переменную с тем же именем внутри функции, она будет **локальной**, и не изменит глобальную:
```python
x = 'awesome'

def myfunc():
    x = 'fantastic'
    print('Python is ' + x)

myfunc()
print('Python is ' + x)
```
### 🌍 Ключевое слово `global`
Обычно переменные, созданные внутри функции, локальны. Но если нужно создать или изменить глобальную переменную, используйте `global`:
```python
def myfunc():
    global x
    x = 'fantastic'

myfunc()
print('Python is ' + x)
```
Вы также можете использовать `global`, чтобы изменить существующую глобальную переменную:
```python
x = 'awesome'

def myfunc():
    global x
    x = 'fantastic'

myfunc()
print('Python is ' + x)
```

### Упражнение 18. Global в условии
Создайте функцию, где если `n > 0`, создаётся `global sign = 'positive'`, иначе `'negative'`.

````python
# Ваш код здесь
n = 1

def if_else():
    global n
    global sign
    if n > 0:
        sign = "positive"
    else: 
        sign = "negative"

if_else()
print(sign)
````
