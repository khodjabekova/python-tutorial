# 🐍 Python: Output Variables

## Вывод переменных (`print()`)
Функция `print()` используется для вывода значений на экран.

```python
x = 'Python is awesome'
print(x)
```
### ✅ Вывод нескольких переменных через запятую
```python
x = 'Python'
y = 'is'
z = 'awesome'
print(x, y, z)
```
Запятая автоматически добавляет пробел между элементами.

### ✅ Объединение строк с `+`
```python
x = 'Python '
y = 'is '
z = 'awesome'
print(x + y + z)
```
Если убрать пробелы, результат будет `Pythonisawesome`.

### ✅ Работа с числами
```python
x = 5
y = 10
print(x + y)
```
### ⚠️ Ошибка при сложении строки и числа
```python
x = 5
y = 'John'
# print(x + y) ❌ TypeError
print(x, y)  # ✅ безопасно
```
Используйте **запятую**, если хотите вывести переменные разных типов.
