## Присваивание нескольким переменным
**Python** позволяет присваивать значения сразу нескольким переменным в одной строке.

### 1. Много значений — многим переменным
```python
x, y, z = 'Orange', 'Banana', 'Cherry'
print(x, y, z)
```
Количество переменных должно **совпадать** с количеством значений.

### 2. Одно значение — многим переменным
```python
x = y = z = 'Orange'
print(x, y, z)
```
Все переменные будут указывать на одно и то же значение.

### 3. Распаковка коллекции (unpacking)
```python
fruits = ['apple', 'banana', 'cherry']
x, y, z = fruits
print(x, y, z)
```
Можно использовать `*` для распаковки нескольких элементов:
```python
a, *b = [1, 2, 3, 4]
print(a)  # 1
print(b)  # [2, 3, 4]
```
