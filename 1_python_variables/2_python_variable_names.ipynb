# 🐍 Python: Variable Names

## Имена переменных в Python
**Переменные** — это имена (ярлыки), которые ссылаются на значения в памяти.

**Правила именования:**
1. Имя должно начинаться с **буквы** или **подчёркивания `_`**.
2. Имя **не может начинаться с цифры**.
3. Допустимы только **буквы, цифры и `_`**.
4. Python чувствителен к регистру: `age`, `Age`, `AGE` — разные переменные.
5. Имя не может совпадать с **ключевыми словами** Python (`if`, `for`, `class` и т.д.).

**Примеры корректных имён:**
```python
myvar = 'John'
my_var = 'John'
_my_var = 'John'
myVar = 'John'
MYVAR = 'John'
myvar2 = 'John'
```
**Некорректные имена:**
```python
2myvar = 'John'  # нельзя начинать с цифры
my-var = 'John'  # дефис запрещён
my var = 'John'  # пробел запрещён
```

### 🧩 Стили для многословных имён
- **camelCase** — каждое слово (кроме первого) с заглавной буквы: `myVariableName`
- **PascalCase** — каждое слово с заглавной: `MyVariableName`
- **snake_case** — слова разделены подчёркиванием: `my_variable_name`
