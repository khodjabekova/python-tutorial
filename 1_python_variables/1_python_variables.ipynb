# 🐍 Python: Переменные 

## Краткая теория
**Переменные** — это имена, указывающие на объекты (значения) в памяти.

**Создание переменной**: переменная создаётся при первом присваивании:
```python
x = 5
name = "John"
```
**Динамическая типизация**: тип значения может меняться при повторном присваивании:
```python
x = 4      # int
x = "Sally"  # str
```
**Приведение типов (casting)**:
```python
x = str(3)   # '3'
y = int(3)   # 3
z = float(3) # 3.0
```
**Проверка типа**: `type(x)` (или лучше `isinstance(x, T)`).

**Строки** — в одинарных или двойных кавычках: `'hi'` и `"hi"` эквивалентны.

**Чувствительность к регистру**: `a` и `A` — разные имена.

**Множественное присваивание / распаковка**:
```python
a, b = 1, 2
first, *middle, last = [1, 2, 3, 4]
```
**Обмен значениями** без временной переменной: `a, b = b, a`.

**Константы (по соглашению)**: Пишем capslock (но это лишь соглашение): `PI = 3.14159`.

**Идентичность и изменяемость**:
- `id(x)` — «адрес» объекта; при переназначении меняется ссылка
- неизменяемые типы: `int`, `float`, `str`, `tuple`, `bool`
- изменяемые: `list`, `dict`, `set`
