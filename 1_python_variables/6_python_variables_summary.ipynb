# 🐍 Python Variables — Summary Practice


## Краткий конспект
- **Variables:** создаются присваиванием; тип динамический; `type(x)`.
- **Variable Names:** буква/`_` в начале; без пробелов/дефисов; `snake_case`.
- **Assign Multiple Values:** `a, b = 1, 2`; одно значение всем: `x = y = 0`; распаковка и `*`.
- **Output Variables:** `print(x, y)`; строки `+`; f-строки; смешанные типы — через запятую.
- **Global Variables:** глобальные вне функций; изменение внутри — через `global`.

### Задание 13. Global • чтение
Глобальная `msg='hello'`; в функции `show()` выведите её без изменения.

```python
# Ваш код здесь
msg = "hello"
def show():
    global msg
    print(msg)

show()
```

### Задание 14. Global • изменение
`flag=False`; в функции `activate()` через `global` сделайте `True`.

```python
# Ваш код здесь
flag = False

def active():
    global flag
    flag = True
    print(flag)

active()
```

### Задание 15. Local vs Global
`x='outer'`; в функции локально `x='inner'`; выведите внутри и снаружи.

```python
# Ваш код здесь
x = "other"

def local():
    global x
    x = "inner"
    print(x)

print(x)

local()
```

### Задание 22. Global • счётчик

```python
# Ваш код здесь
counter = 0
def inc():
    global counter
    counter +=1
    print(counter)

print(counter)
inc()
inc()
inc()
```

### Задание 23. Распаковка в цикле
